### Loss Slope

$$
\frac{\partial L}{\partial m}
=
-2 \sum_{i=1}^{n} X_i\left(Y_i-mX_i-b\right)
$$

### Update Intercept

$$
b_{\text{new}}
=
b_{\text{old}}
-
\text{learning\_rate}
\times
\text{Loss Slope}
$$

```python
class GDRegressor:
    def __init__(self,learning_rate, epochs):
        self.m = 33.70
        self.b = -120
        self.lr = learning_rate
        self.epochs = epochs
        
    def fit(self,X,y):
        # Calculate the b using Gradient Descent
        
        for i in range(self.epochs):
            loss_slope = -2 * np.sum(y - self.m * X.ravel() - self.b)
            
            self.b = self.b - (self.lr * loss_slope)
            
            print(loss_slope, self.b)
            
        print(f"b = {self.b}")
            
```

```python
gd = GDRegressor(0.001,100)
```

**Initially, we used learning rates of 0.1 and 0.01 with 10 epochs. However, the model's parameter kept oscillating between approximately 111 and 120 instead of converging. This indicates that the learning rate was too large for the current optimization landscape. To allow smaller, more stable updates, we reduced the learning rate to 0.001 and increased the number of epochs to 100.**
