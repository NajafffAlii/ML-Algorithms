# If data is sort of linear and not Completely Linear it is because of Stochastic Errors

What are Stochastic Errors?

A stochastic error is the random part of the data that cannot be explained by the model. It represents the unpredictable variation caused by factors we cannot measure or control.

# In a regression equation:

Y = β0 + β1X + ε

Y = Target (dependent variable)
X = Feature (independent variable)
β0,β1 = Model parameters
ε = Stochastic (random) error

The term ε captures everything affecting Y that is not included in the model.

# The line we draw in Regression is called Best Fit Line

# To Find mx + b:
    There are two ways:
    1) Closed Form Solution:
     --> OLS(Ordinary Least Squares), Direct Formula
     --> For small number of Dimensions

    2) Open Form Solution:
    --> Gradient Discent

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
class LinearRegression:
    def __init__(self):
        self.m = None
        self.b = None
        
    def Fit(self, X_train, Y_train):
        num = 0
        den = 0
        
        for i in range(X_train.shape[0]):
            num +=  (X_train[i] - X_train.mean()) * (Y_train[i] - Y_train.mean())
            den += (X_train[i] - X_train.mean()) ** 2
            
        self.m = num / den
        self.b = Y_train.mean() - (self.m * X_train.mean())
        
        print(self.m)
        print(self.b)
        
    def Predict(self, X_test):
        return self.m * X_test + self.b
```
