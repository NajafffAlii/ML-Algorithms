# Sigmoid Function

Implemented the **Sigmoid Function from scratch** using Python.

[
\sigma(x)=\frac{1}{1+e^{-x}}
]

It converts any input into a value between **0 and 1**. Tested it with different values and plotted the sigmoid curve using **NumPy** and **Matplotlib**.


# Negative inputs → probability closer to 0
# Positive inputs → probability closer to 1.

```python
import math
import numpy as np 
import matplotlib.pyplot as plt
```

```python
def Sigmoid(x):
    return 1 / (1+ math.exp(-x))
```

```python
x = np.linspace(-10,10,100)
```

```python
y = np.array([Sigmoid(value) for value in x])
```

```python
plt.plot(x,y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Sigmoid Function")

plt.grid()
plt.show()

# Curve squashes any input from −∞,+∞ into the range 0,1.
```

```python
# For testing Purpose:

values = [-5, -2, -1, 0, 1, 2, 5]

for i in values:
    print(f"x = {i}, Sigmoid = {Sigmoid(i): .4f}")
```
