```python
import numpy as np
```

```python
def ReLu(x):
    return np.maximum(0,x)
```

```python
x = np.array([-5,-2,-1,0,1,2,5])
```

```python
print("Original Array:\n",x)
print("Array after ReLu:\n",ReLu(x))
```

# ReLu Derivative

```python
def ReLu_Derivative(x):
    return (x > 0).astype(float)
```

```python
x = np.array([-5,-2,-1,0,1,2,5])
print("ReLu Derivative:\n", ReLu_Derivative(x))
```
