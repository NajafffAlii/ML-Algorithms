## Binary Cross Entropy (BCE) From Scratch

Implemented Binary Cross Entropy from scratch to measure the loss in binary classification.

- Used `y_true` for actual labels and `y_pred` for predicted probabilities from Sigmoid.
- Used `epsilon` and `np.clip()` to prevent `log(0)`.
- Applied the BCE formula to calculate the loss for each sample.
- Used `np.mean()` to calculate the average loss.

### Flow
Model Output → Sigmoid → Predicted Probability → BCE → Loss

This implementation produced a BCE loss of approximately **0.2271** for our test predictions.

```python
import numpy as np
```

```python
def BCE(y_true,y_pred):
    epsilon = 1e-15

    y_pred = np.clip(y_pred, epsilon, 1 - epsilon)

    # Now implement BCE Formula: −[ylog(p)+(1−y)log(1−p)]
    
    loss = -(y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))

    return np.mean(loss)
```

```python
y_true = np.array([1, 0, 1, 1])
y_pred = np.array([0.9, 0.2, 0.8, 0.7])
```

```python
loss = BCE(y_true, y_pred)

print("BCE Loss:", loss)
```
