## Tanh — Implementation Summary

Tanh (Hyperbolic Tangent) maps any input to the range **(-1, 1)** and is a zero-centered activation function.

### Formula
tanh(x) = e^x - e^(-x) / e^x + e^(-x)

### Derivative
tanh'(x) = 1-tanh^2(x)

### From Scratch
We implemented both the forward function and its derivative using NumPy, without using `np.tanh()`.

- **Forward pass:** calculates the Tanh output using its exponential formula.
- **Derivative:** calculates 1-tanh^2(x).
- **Testing:** verified the outputs for negative, zero, and positive inputs.

This showed that Tanh is **zero-centered**, but its gradient becomes very small when the output approaches **-1 or 1**, which can lead to the **vanishing gradient problem**.

```python
import numpy as np
```

```python
def Tanh(x):
    return ((np.exp(x) - np.exp(-x)) / (np.exp(x) + np.exp(-x)))
```

```python
x = np.array([-2, -1, 0, 1, 2])

print(Tanh(x))
```

```python
def Derivative_of_Tanh(x):
    return (1 - Tanh(x) ** 2)
```
