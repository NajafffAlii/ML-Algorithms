# Numerically stable Softmax Implementation 

We implemented **Softmax from scratch** using NumPy.

1. **Shift values for numerical stability:**
   Subtract the maximum value to prevent exponential overflow.

   ```python
   shifted_x = x - np.max(x)
   ```

2. **Calculate exponentials:**
   Convert each value into a positive exponential value.

   ```python
   exp_x = np.exp(shifted_x)
   ```

3. **Normalize into probabilities:**
   Divide each exponential by their total sum so that all probabilities add up to 1.

   ```python
   return exp_x / np.sum(exp_x)
   ```

**Result:** Softmax converts logits into probabilities that sum to **1**.


````python
import numpy as np
````

````python
def SoftMax(x):
    shifted_x = x - np.max(x)
    exp_x = np.exp(shifted_x)
    return  exp_x / np.sum(exp_x)
````

````python
x = np.array([2.0,2.0,2.0])
````

````python
print("Input: ", x)
print("SoftMax: ", SoftMax(x))
print("Sum: ", np.sum(SoftMax(x)))
````

````python
x = np.array([1000, 999, 998])

print(SoftMax(x))
print(np.sum(SoftMax(x)))
````
